```python
import os
from os import path
import sys
import matplotlib.pyplot as plt
import numpy as np
```

```python
q = 2
L = 32
T = 1.15

path = './configs_wolff_eq20000f100/'
filename = os.path.join(path, "potts_configs_q{}L{}T{:.3f}.npy".format(q,L,T))
configs = np.load(filename)

print('analyze configuration file:', filename)
print(configs.shape)
```

```python
# if you want to use self-defined colors
import matplotlib.colors as mcolors
# Define discrete colors
colors = ['red', 'blue'] # ['red', 'green', 'blue', 'orange']  # one color per spin
cmap = mcolors.ListedColormap(colors)
# Normalization: map integer values to colors
bounds = np.arange(q+1) - 0.5  # centers integers in color bins
norm = mcolors.BoundaryNorm(bounds, cmap.N)
```

## Show 256 consecutive configurations

```python
fig = plt.figure(figsize=(16, 16)) 
ishift = 1000
for i in range(256):  # 16 × 16 grid
    ax = fig.add_subplot(16, 16, i + 1)
    ax.imshow(configs[ishift+i].reshape(L, L), cmap='tab10')
    #ax.imshow(configs[i].reshape(L, L), cmap=cmap)
    # ax.axis('off')
    ax.set_xticks([])
    ax.set_yticks([])

plt.tight_layout()
plt.show()
```

## Show 256 randomly selected configurations

```python
indices = np.random.choice(len(configs), size=256, replace=False)

fig = plt.figure(figsize=(16, 16)) 

for i, idx in enumerate(indices):
    ax = fig.add_subplot(16, 16, i + 1)
    ax.imshow(configs[idx].reshape(L, L), cmap='tab10') # cmap=cmap
    ax.set_xticks([])
    ax.set_yticks([])
    # ax.axis('off')
plt.tight_layout()
plt.show()
```
